```python
#Imports and project setup
from pathlib import Path
import pandas as pd
import numpy as np

# Finding project root
cwd = Path.cwd()

if cwd.name.lower() == "notebooks":
    PROJECT_ROOT = cwd.parent
else:
    PROJECT_ROOT = cwd

print("NOTEBOOK 06 — UPAg UNIT HARMONIZATION")

print("Current directory :", cwd)
print("Project root      :", PROJECT_ROOT)

# Important directories
DATA_RAW = PROJECT_ROOT / "data" / "raw"
DATA_PROCESSED = PROJECT_ROOT / "data" / "processed"

print("\nRaw data path      :", DATA_RAW)
print("Processed path     :", DATA_PROCESSED)

print("\nSetup complete.")
```

```python
#Locating upag csv files
print("SEARCHING FOR UPAg CSV FILES")

upag_files = []

for path in DATA_RAW.rglob("*.csv"):
    if "upag" in str(path).lower():
        upag_files.append(path)

print(f"\nUPAg CSV files found: {len(upag_files)}")

for i, path in enumerate(upag_files, 1):
    print(f"{i}. {path}")
```

```python
#Loading and inspecting upag raw data
print("LOADING UPAg RAW DATA")

upag = {}

for path in upag_files:
    crop = path.parent.name.capitalize()
    
    df = pd.read_csv(path, low_memory=False)
    upag[crop] = df
    
    print(f"\n{'-' * 70}")
    print(f"CROP: {crop}")
    print(f"Shape: {df.shape}")
    print(f"Columns: {list(df.columns)}")
    print("\nSample:")
    display(df.head(3))

print("UPAg FILES LOADED SUCCESSFULLY")
```

```python
#Loading and reshaping upag data
from pathlib import Path
import pandas as pd

print("LOADING AND RESHAPING UPAg DATA")

# Finding project root
PROJECT_ROOT = Path.cwd().parent

# If notebook is inside Dump/notebooks
if not (PROJECT_ROOT / "data").exists():
    PROJECT_ROOT = Path.cwd().parent.parent

print("Project root:", PROJECT_ROOT)

# Finding UPAg files recursively
upag_files = list(
    (PROJECT_ROOT / "data" / "raw" / "upag").rglob("*.csv")
)

print("UPAg files found:", len(upag_files))

for p in upag_files:
    print(" -", p)

#Identifying crop files
crop_paths = {}

for path in upag_files:

    name = path.name.lower()

    if "rice" in name:
        crop_paths["Rice"] = path

    elif "wheat" in name:
        crop_paths["Wheat"] = path

    elif "maize" in name:
        crop_paths["Maize"] = path

    elif "urad" in name:
        crop_paths["Urad"] = path


print("\nCrop files identified:")
for crop, path in crop_paths.items():
    print(f"✓ {crop:10s} -> {path}")

# Crop-specific column prefixes
prefixes = {
    "Rice": "rice",
    "Wheat": "wheat",
    "Maize": "maz",
    "Urad": "urad"
}

# Reshaping
upag_long = {}

for crop, path in crop_paths.items():

    print("\n" + "-" * 70)
    print("Processing:", crop)

    df = pd.read_csv(path, low_memory=False)

    prefix = prefixes[crop]

    rows = []

    for _, row in df.iterrows():

        for year, suffix in [
            ("2024-25", "25"),
            ("2023-24", "24"),
            ("2022-23", "23")
        ]:

            rows.append({
                "year": year,
                "state": row["state"],
                "district": row["district"],
                "lgd_distcode": row["lgd_distcode"],
                "lgd_statecode": row["lgd_statecode"],
                "crop": crop,

                "area": row[f"{prefix}area{suffix}"],
                "production": row[f"{prefix}prod{suffix}"],
                "yield": row[f"{prefix}yld{suffix}"]
            })

    upag_long[crop] = pd.DataFrame(rows)

    print("Raw rows :", len(df))
    print("Long rows:", len(upag_long[crop]))

print("RESHAPING COMPLETE")

print("\nFinal shapes:")

for crop, df in upag_long.items():
    print(f"{crop:10s} -> {df.shape}")

print("\nSample:")
display(upag_long["Rice"].head())
```

```python
#Inspecting UPAg units and comparing with Notebook 05 standard
print("UPAg UNIT INSPECTION")

for crop, df in upag_long.items():
    print(f"\n{crop}")
    print("-" * 60)

    print("Area:")
    print(df["area"].describe())

    print("\nProduction:")
    print(df["production"].describe())

    print("\nYield:")
    print(df["yield"].describe())

print("UPAg CURRENT UNIT INTERPRETATION")

print("""
UPAg area values  : million hectares (likely)
UPAg production   : million tonnes (likely)
UPAg yield        : kg/hectare

Notebook 05 standard:
Area              : hectare
Production        : tonnes
Yield              : tonnes/hectare
""")
```

```python
print("UPAg RAW DATA — UNIT INSPECTION")

from pathlib import Path
import pandas as pd

# Finding project root from the notebook location
PROJECT_ROOT = Path.cwd().parent

# Searching for UPAg CSV files
upag_files = list(
    PROJECT_ROOT.rglob("*_upag_raw.csv")
)

print(f"UPAg files found: {len(upag_files)}")

for path in upag_files:

    print("\n" + "-" * 70)
    print(f"FILE: {path}")
    print("-" * 70)

    raw = pd.read_csv(path, low_memory=False)

    print("Shape:", raw.shape)

    measure_cols = [
        c for c in raw.columns
        if any(word in c.lower()
               for word in ["area", "prod", "yld", "yield"])
    ]

    print("\nMeasurement columns:")
    print(measure_cols)

    print("\nSample values:")
    print(
        raw[measure_cols]
        .head(3)
        .to_string(index=False)
    )
```

```python
print("UPAg UNIT HARMONIZATION")

upag_harmonized = {}

for crop, df in upag_long.items():

    temp = df.copy()

    # UPAg area and production are stored in million units
    # Converting million hectares -> hectares
    # Converting million tonnes -> tonnes
    temp["area_ha"] = temp["area"] * 1_000_000
    temp["production_tonnes"] = temp["production"] * 1_000_000

    # UPAg yield is kg/hectare
    # Converting kg/hectare -> tonnes/hectare
    temp["yield_reported"] = temp["yield"] / 1_000

    # Keeping standardized columns
    temp = temp[
        [
            "year",
            "state",
            "district",
            "lgd_distcode",
            "lgd_statecode",
            "crop",
            "area_ha",
            "production_tonnes",
            "yield_reported"
        ]
    ]

    upag_harmonized[crop] = temp

    print(f"\n{crop}")
    print("-" * 60)
    print("Rows:", len(temp))
    print("Area unit       : hectare")
    print("Production unit : tonnes")
    print("Yield unit      : tonnes/hectare")

print("UNIT HARMONIZATION COMPLETE")
```

```python
print("VALIDATING UPAg UNIT CONVERSION")

validation_results = []

for crop, df in upag_harmonized.items():

    temp = df.copy()

    # Calculating yield independently
    temp["yield_calculated"] = (
        temp["production_tonnes"] / temp["area_ha"]
    )

    # Comparing only valid non-zero records
    valid = temp[
        (temp["area_ha"] > 0) &
        (temp["production_tonnes"].notna()) &
        (temp["yield_reported"].notna())
    ].copy()

    # Difference
    valid["yield_difference"] = (
        valid["yield_reported"] -
        valid["yield_calculated"]
    )

    validation_results.append({
        "crop": crop,
        "rows": len(temp),
        "valid_rows": len(valid),
        "mean_difference": valid["yield_difference"].mean(),
        "max_abs_difference": valid["yield_difference"].abs().max()
    })

validation_df = pd.DataFrame(validation_results)

print(validation_df.to_string(index=False))

print("VALIDATION COMPLETE")
```

```python
print("UPAg YIELD DIFFERENCE ANALYSIS")

difference_summary = []

for crop, df in upag_harmonized.items():

    temp = df.copy()

    temp["yield_calculated"] = (
        temp["production_tonnes"] / temp["area_ha"]
    )

    valid = temp[
        (temp["area_ha"] > 0) &
        temp["production_tonnes"].notna() &
        temp["yield_reported"].notna()
    ].copy()

    valid["abs_difference"] = (
        valid["yield_reported"] -
        valid["yield_calculated"]
    ).abs()

    difference_summary.append({
        "crop": crop,
        "valid_rows": len(valid),
        "difference_<=_0.1": (valid["abs_difference"] <= 0.1).sum(),
        "difference_<=_0.5": (valid["abs_difference"] <= 0.5).sum(),
        "difference_>_0.5": (valid["abs_difference"] > 0.5).sum(),
        "difference_>_1.0": (valid["abs_difference"] > 1.0).sum()
    })

difference_df = pd.DataFrame(difference_summary)

print(difference_df.to_string(index=False))

print("INTERPRETATION")
print("""
Reported UPAg yield will be retained as the authoritative yield field.
Calculated yield is used only for quality validation.

No yield values will be overwritten.
""")
```

```python
#Geographic + Duplicate + Missing-Value Validation
print("UPAg FINAL DATA VALIDATION")

for crop, df in upag_long.items():

    print(f"\n{crop}")
    print("-" * 60)

    print("Rows:", len(df))
    print("States:", df["state"].nunique())
    print("Districts:", df["district"].nunique())
    print("Years:", df["year"].nunique())

    missing = df[["state", "district", "area",
                  "production", "yield"]].isna().sum()

    duplicates = df.duplicated(
        subset=["year", "state", "district", "crop"]
    ).sum()

    print("Missing values:")
    print(missing.to_dict())

    print("Duplicate records:", duplicates)

print("FINAL VALIDATION COMPLETE")
```

```python
#Combining the 4 crops
print("CREATING FINAL UPAg HARMONIZED DATASET")

upag_harmonized = pd.concat(
    upag_long.values(),
    ignore_index=True
)

upag_harmonized = upag_harmonized.sort_values(
    ["crop", "year", "state", "district"]
).reset_index(drop=True)

print("Final shape:", upag_harmonized.shape)
print("Crops:", upag_harmonized["crop"].unique())
print("Columns:", list(upag_harmonized.columns))

upag_harmonized.head()
```
