```python
from pathlib import Path
import pandas as pd

# Current notebook directory
NOTEBOOK_DIR = Path.cwd()

# Project root: ...\Dump
PROJECT_ROOT = NOTEBOOK_DIR.parent

# Data directories
RAW = PROJECT_ROOT / "data" / "raw"
PROCESSED = PROJECT_ROOT / "data" / "processed"

# Input files
DES_FILE = PROCESSED / "des" / "des_harmonized.csv"
UPAG_FILE = PROCESSED / "upag_harmonized.csv"

print("Current directory :", NOTEBOOK_DIR)
print("Project root      :", PROJECT_ROOT)
print("Processed path    :", PROCESSED)

print("\nExpected input files:")
print("DES  :", DES_FILE)
print("UPAg :", UPAG_FILE)

print("\nSetup complete.")
```

```python
print("LOCATING HARMONIZED DATASETS")

# Searching the complete processed-data directory
all_csvs = list(PROJECT_ROOT.rglob("*.csv"))

print(f"Processed CSV files found: {len(all_csvs)}")

for f in all_csvs:
    print(" -", f.relative_to(PROJECT_ROOT))

# Locating DES and UPAg files by filename
des_matches = [
    f for f in all_csvs
    if "des" in f.name.lower() and "harmon" in f.name.lower()
]

upag_matches = [
    f for f in all_csvs
    if "upag" in f.name.lower() and "harmon" in f.name.lower()
]

print("\nDES candidates:")
for f in des_matches:
    print(" ✓", f)

print("\nUPAg candidates:")
for f in upag_matches:
    print(" ✓", f)
```

```python
print("INSPECTING DES AND UPAg CANDIDATE FILES")

# DES candidates: processed crop-yield datasets
des_candidates = [
    f for f in all_csvs
    if "processed" in str(f).lower()
    and "crop_yield" in str(f).lower()
    and "des" in f.name.lower()
]

# UPAg candidates
upag_candidates = [
    f for f in all_csvs
    if "upag" in f.name.lower()
    and "processed" in str(f).lower()
]

print("\nDES CANDIDATES")
print("-" * 80)

for f in des_candidates:
    temp = pd.read_csv(f, nrows=3)
    print(f"\nFile: {f.relative_to(PROJECT_ROOT)}")
    print("Columns:", list(temp.columns))

print("\nUPAg CANDIDATES")
print("-" * 80)

for f in upag_candidates:
    temp = pd.read_csv(f, nrows=3)
    print(f"\nFile: {f.relative_to(PROJECT_ROOT)}")
    print("Columns:", list(temp.columns))
```

```python
print("LOADING DES AND UPAg HARMONIZED DATA")

# Actual DES dataset
DES_FILE = PROJECT_ROOT / "data" / "processed" / "crop_yield" / "des_apy_2013_14_2022_23.csv"

# Actual UPAg harmonized dataset
UPAG_FILE = PROJECT_ROOT / "data" / "processed" / "upag_harmonized.csv"

des = pd.read_csv(DES_FILE)
upag = pd.read_csv(UPAG_FILE)

print("\nDES")
print("Shape:", des.shape)
print("Columns:", list(des.columns))

print("\nUPAg")
print("Shape:", upag.shape)
print("Columns:", list(upag.columns))

print("\n✓ Both datasets loaded successfully")
```

```python
print("COMPARING DES AND UPAg COVERAGE")

print("\nDES crops:")
print(sorted(des["crop"].dropna().unique()))

print("\nUPAg crops:")
print(sorted(upag["crop"].dropna().unique()))

print("\nDES year range:")
print(des["year"].min(), "to", des["year"].max())

print("\nUPAg year range:")
print(upag["year"].min(), "to", upag["year"].max())
```

```python
print("IDENTIFYING COMMON CROPS AND OVERLAPPING YEARS")

common_crops = sorted(set(des["crop"]) & set(upag["crop"]))
common_years = sorted(set(des["year"]) & set(upag["year"]))

print("\nCommon crops:")
for crop in common_crops:
    print("✓", crop)

print("\nOverlapping years:")
for year in common_years:
    print("✓", year)

print("\nCommon crop count :", len(common_crops))
print("Common year count :", len(common_years))
```

```python
print("NORMALIZING YEAR FORMATS")

def normalize_year(x):
    x = str(x).strip()
    
    if "-" not in x:
        return x
    
    start, end = x.split("-", 1)
    
    # Converting YYYY-YY → YYYY-YYYY
    if len(start) == 4 and len(end) == 2:
        end = start[:2] + end
    
    return f"{start}-{end}"

des["year_normalized"] = des["year"].apply(normalize_year)
upag["year_normalized"] = upag["year"].apply(normalize_year)

print("DES years   :", sorted(des["year_normalized"].unique()))
print("UPAg years  :", sorted(upag["year_normalized"].unique()))

common_years = sorted(
    set(des["year_normalized"]) & set(upag["year_normalized"])
)

print("\nCommon years:", common_years)
print("Common year count:", len(common_years))
```

```python
print("NORMALIZING YEARS AND CREATING COMPARISON SUBSETS")

def normalize_year(x):
    x = str(x).strip()
    if "-" in x:
        a, b = x.split("-")
        if len(b) == 2:
            b = a[:2] + b
        return f"{a}-{b}"
    return x

des["year"] = des["year"].apply(normalize_year)
upag["year"] = upag["year"].apply(normalize_year)

common_crops = ["Maize", "Rice", "Urad", "Wheat"]
common_year = "2022-2023"

des_common = des[
    des["crop"].isin(common_crops) &
    (des["year"] == common_year)
].copy()

upag_common = upag[
    upag["crop"].isin(common_crops) &
    (upag["year"] == common_year)
].copy()

print("DES comparison rows  :", len(des_common))
print("UPAg comparison rows :", len(upag_common))

print("\nDES crop counts:")
print(des_common["crop"].value_counts().sort_index())

print("\nUPAg crop counts:")
print(upag_common["crop"].value_counts().sort_index())

print("\nComparison subsets created successfully.")
```

```python
print("AGGREGATING DES SEASONAL DATA TO ANNUAL DISTRICT LEVEL")

# Annual aggregation of DES (sum area & production across seasons)
des_annual = (
    des_common
    .groupby(
        ["year", "state", "district", "crop"],
        as_index=False
    )
    .agg({
        "area_ha": "sum",
        "production_tonnes": "sum"
    })
)

# Recalculating annual yield
des_annual["yield_kg_ha"] = (
    des_annual["production_tonnes"] /
    des_annual["area_ha"]
) * 1000

# Renaming columns to match UPAg naming
des_annual = des_annual.rename(columns={
    "area_ha": "area",
    "production_tonnes": "production",
    "yield_kg_ha": "yield"
})

print("DES seasonal rows :", len(des_common))
print("DES annual rows   :", len(des_annual))

print("\nCrop-wise annual records:")
print(des_annual["crop"].value_counts().sort_index())

print("\nSample annual DES data:")
display(des_annual.head())
```

```python
print("STANDARDIZING STATE AND DISTRICT NAMES")

def clean_location(series):
    return (
        series.astype(str)
        .str.strip()
        .str.replace(r"\s+", " ", regex=True)
        .str.lower()
    )

for df in [des_annual, upag_common]:
    df["state_key"] = clean_location(df["state"])
    df["district_key"] = clean_location(df["district"])

print("DES unique states     :", des_annual["state_key"].nunique())
print("UPAg unique states    :", upag_common["state_key"].nunique())
print("DES unique districts  :", des_annual["district_key"].nunique())
print("UPAg unique districts :", upag_common["district_key"].nunique())

print("\nGeographic keys created successfully.")
```

```python
print("MATCHING DES AND UPAg RECORDS")

match_keys = ["year", "state_key", "district_key", "crop"]

des_keys = des_annual[match_keys].drop_duplicates()
upag_keys = upag_common[match_keys].drop_duplicates()

matched_keys = des_keys.merge(
    upag_keys,
    on=match_keys,
    how="inner"
)

print("DES annual records :", len(des_keys))
print("UPAg records       :", len(upag_keys))
print("Exact matches      :", len(matched_keys))

print("\nMatches by crop:")
print(matched_keys["crop"].value_counts().sort_index())

print("\nExact geographic matching completed.")
```

```python
print("CREATING DES–UPAg HARMONIZED COMPARISON DATA")

des_compare = des_annual.merge(
    upag_common[
        ["year", "state_key", "district_key", "crop",
         "area", "production", "yield"]
    ],
    on=["year", "state_key", "district_key", "crop"],
    how="inner",
    suffixes=("_des", "_upag")
)

print("Matched comparison rows :", len(des_compare))
print("Crops                  :", des_compare["crop"].unique())

print("\nColumns:")
print(des_compare.columns.tolist())

display(des_compare.head())
```

```python
print("CALCULATING DES–UPAg DIFFERENCES")

for metric in ["area", "production", "yield"]:
    des_compare[f"{metric}_diff"] = (
        des_compare[f"{metric}_upag"] -
        des_compare[f"{metric}_des"]
    )

    des_compare[f"{metric}_diff_pct"] = (
        des_compare[f"{metric}_diff"] /
        des_compare[f"{metric}_des"].replace(0, pd.NA)
    ) * 100

print("Difference calculations completed.")

display(des_compare[
    ["year", "state", "district", "crop",
     "area_des", "area_upag", "area_diff_pct",
     "production_des", "production_upag", "production_diff_pct",
     "yield_des", "yield_upag", "yield_diff_pct"]
].head())
```

```python
print("DES–UPAg CROP-WISE COMPARISON SUMMARY")

summary = (
    des_compare
    .groupby("crop")
    .agg(
        matched_rows=("crop", "size"),
        mean_area_diff_pct=("area_diff_pct", "mean"),
        median_area_diff_pct=("area_diff_pct", "median"),
        mean_production_diff_pct=("production_diff_pct", "mean"),
        median_production_diff_pct=("production_diff_pct", "median"),
        mean_yield_diff_pct=("yield_diff_pct", "mean"),
        median_yield_diff_pct=("yield_diff_pct", "median")
    )
    .reset_index()
)

display(summary.round(2))
```

```python
print("DES–UPAg MATCHING SUMMARY")

total_des = len(des_annual)
matched = len(des_compare)

print(f"DES annual records : {total_des}")
print(f"Matched records    : {matched}")
print(f"Unmatched records  : {total_des - matched}")
print(f"Match rate         : {matched / total_des * 100:.2f}%")

print("\nMatched records by crop:")
print(
    des_compare["crop"]
    .value_counts()
    .sort_index()
)

print("\nComparison period : 2022-2023")
print("Comparison crops  : Maize, Rice, Urad, Wheat")
```

```python
print("CLASSIFYING UPAg OBSERVATIONS")

for col in ["area_upag", "production_upag", "yield_upag"]:
    des_compare[col + "_status"] = des_compare[col].apply(
        lambda x: "missing" if pd.isna(x)
        else "zero" if x == 0
        else "valid"
    )

status_summary = des_compare.groupby("crop")[
    ["area_upag_status", "production_upag_status", "yield_upag_status"]
].agg(lambda x: x.value_counts().to_dict())

display(status_summary)

print("\nObservation classification complete.")
```

```python
print("CALCULATING VALID DES–UPAg DIFFERENCES")

for metric in ["area", "production", "yield"]:
    des_col = metric + "_des"
    upag_col = metric + "_upag"
    diff_col = metric + "_diff_pct"

    valid = (
        des_compare[des_col].notna() &
        des_compare[upag_col].notna() &
        (des_compare[upag_col] > 0)
    )

    des_compare.loc[valid, diff_col] = (
        (des_compare.loc[valid, upag_col] -
         des_compare.loc[valid, des_col])
        / des_compare.loc[valid, des_col].replace(0, pd.NA)
    ) * 100

print("Valid difference calculations completed.")

display(
    des_compare[
        ["crop",
         "area_diff_pct",
         "production_diff_pct",
         "yield_diff_pct"]
    ].describe()
)
```

```python
print("DES vs UPAg VALUE SCALE CHECK")

scale_check = des_compare.groupby("crop").agg(
    des_area_median=("area_des", "median"),
    upag_area_median=("area_upag", "median"),
    des_production_median=("production_des", "median"),
    upag_production_median=("production_upag", "median"),
    des_yield_median=("yield_des", "median"),
    upag_yield_median=("yield_upag", "median")
)

display(scale_check)
```

```python
print("VALID YIELD COMPARISON")

valid_yield = (
    des_compare["yield_des"].notna() &
    des_compare["yield_upag"].notna()
)

des_compare["yield_diff_pct"] = (
    (des_compare["yield_upag"] - des_compare["yield_des"])
    / des_compare["yield_des"].replace(0, pd.NA)
) * 100

yield_summary = des_compare[valid_yield].groupby("crop").agg(
    matched_rows=("yield_des", "size"),
    mean_yield_diff_pct=("yield_diff_pct", "mean"),
    median_yield_diff_pct=("yield_diff_pct", "median"),
    mean_des_yield=("yield_des", "mean"),
    mean_upag_yield=("yield_upag", "mean")
).reset_index()

display(yield_summary.round(4))

print("\nArea and production scale discrepancy retained for documentation.")
print("No arbitrary unit conversion applied.")
```

```python
print("FINAL DES–UPAg COMPARISON SUMMARY")

final_summary = yield_summary.copy()

final_summary["match_rate_pct"] = (
    final_summary["matched_rows"]
    / des_compare.groupby("crop").size().values
) * 100

display(final_summary.round(4))

print("\nComparison year : 2022-2023")
print("Comparison crops: Maize, Rice, Urad, Wheat")
print("Area/production: scale discrepancy retained; no arbitrary conversion applied.")
print("Yield comparison: completed successfully.")
```
