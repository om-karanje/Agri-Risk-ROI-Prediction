```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
```

```python
RAW_PATH = "../data/raw/des/des_apy_raw.csv"

df = pd.read_csv(RAW_PATH)

print("Shape:", df.shape)
df.head()
```

```python
#Inspecting Schema
print("Columns:")
for i, col in enumerate(df.columns, start=1):
    print(f"{i}. {col}")
```

```python
#Basic Info
df.info()
```

```python
df.describe(include="all")
```

```python
#Year coverage
year_counts = (
    df.groupby("year")
      .size()
      .reset_index(name="records")
      .sort_values("year")
)

display(year_counts)
```

```python
#State coverage
print("Number of states/UTs:",
      df["state_name"].nunique())

print("\nStates/UTs:")
print(sorted(df["state_name"].dropna().unique()))
```

```python
#District Coverage
print("Number of districts:",
      df["district_name"].nunique())

print("Missing district names:",
      df["district_name"].isna().sum())
```

```python
#Crop coverage
crop_summary = (
    df.groupby(["crop_name", "crop_type"])
      .agg(
          years=("year", "nunique"),
          records=("year", "size"),
          districts=("district_name", "nunique")
      )
      .sort_values("records", ascending=False)
)

display(crop_summary)
```

```python
#Season Coverage
season_summary = (
    df.groupby("season")
      .agg(
          records=("year", "size"),
          crops=("crop_name", "nunique"),
          districts=("district_name", "nunique")
      )
      .sort_values("records", ascending=False)
)

display(season_summary)
```

```python
#Missing value audit
missing = (
    df.isna()
      .sum()
      .reset_index()
      .rename(columns={
          "index": "column",
          0: "missing_count"
      })
)

missing["missing_percent"] = (
    missing["missing_count"] / len(df) * 100
)

display(
    missing.sort_values(
        "missing_count",
        ascending=False
    )
)
```

```python
#Duplicate check
key_columns = [
    "state_name",
    "district_name",
    "crop_name",
    "season",
    "year"
]

duplicates = df.duplicated(
    subset=key_columns,
    keep=False
)

print("Duplicate records:",
      duplicates.sum())
```

```python
#Checking units
print("Area units:")
print(df["area_unit"].value_counts(dropna=False))

print("\nProduction units:")
print(df["production_unit"].value_counts(dropna=False))

print("\nYield units:")
print(df["yield_unit"].value_counts(dropna=False))
```

HISOTRICAL CROP SUBSET OF THE MAIN DATASET

```python
START_YEAR = "2013-2014"
END_YEAR = "2022-2023"

historical_df = df[
    df["year"].between(
        START_YEAR,
        END_YEAR
    )
].copy()

print("Historical dataset shape:",
      historical_df.shape)

print(
    historical_df["year"]
    .value_counts()
    .sort_index()
)
```

```python
#Target crops for initial version
TARGET_CROPS = [
    "Rice",
    "Wheat",
    "Maize",
    "Jowar",
    "Bajra",
    "Ragi",
    "Soyabean",
    "Groundnut",
    "Cotton",
    "Gram",
    "Arhar/Tur",
    "Urad",
    "Sugarcane"
]
```

```python
crop_df = historical_df[
    historical_df["crop_name"].isin(TARGET_CROPS)
].copy()

print("Selected crop dataset:",
      crop_df.shape)

print(
    crop_df["crop_name"]
    .value_counts()
)
```

```python
#Standardizing numeric columns
numeric_columns = [
    "area",
    "production",
    "yield"
]

for col in numeric_columns:
    crop_df[col] = pd.to_numeric(
        crop_df[col],
        errors="coerce"
    )

crop_df[numeric_columns].describe()
```

```python
#Creating standardized units
crop_df["area_ha"] = crop_df["area"]

crop_df["production_tonnes"] = crop_df["production"]

crop_df["yield_tonnes_ha"] = crop_df["yield"]

crop_df["yield_kg_ha"] = (
    crop_df["yield_tonnes_ha"] * 1000
)

crop_df["yield_quintal_acre"] = (
    crop_df["yield_tonnes_ha"] * 10 / 2.47105
)
```

```python
#Adding the source field
crop_df["data_source"] = "DES_APY"
```

```python
#Selecting clean columns
FINAL_COLUMNS = [
    "state_name",
    "state_code",
    "district_name",
    "district_code",
    "crop_name",
    "crop_code",
    "crop_type",
    "season",
    "year",
    "area_ha",
    "production_tonnes",
    "yield_tonnes_ha",
    "yield_kg_ha",
    "yield_quintal_acre",
    "data_source"
]

crop_clean = crop_df[FINAL_COLUMNS].copy()
```

```python
#Renaming them
crop_clean = crop_clean.rename(columns={
    "state_name": "state",
    "district_name": "district",
    "crop_name": "crop_source_name"
})
```

```python
#Adding temporary standard crop name
crop_clean["crop"] = crop_clean["crop_source_name"]
```

```python
#Arranging the columns
crop_clean = crop_clean[
    [
        "state",
        "state_code",
        "district",
        "district_code",
        "crop",
        "crop_source_name",
        "crop_code",
        "crop_type",
        "season",
        "year",
        "area_ha",
        "production_tonnes",
        "yield_tonnes_ha",
        "yield_kg_ha",
        "yield_quintal_acre",
        "data_source"
    ]
]
```

```python
#Running the first quality checks
print("Final shape:", crop_clean.shape)

print("\nMissing values:")
print(
    crop_clean.isna().sum()
    .sort_values(ascending=False)
)

print("\nDuplicates:")
print(
    crop_clean.duplicated(
        subset=[
            "state",
            "district",
            "crop",
            "season",
            "year"
        ]
    ).sum()
)
```

```python
#Checking for impossible values
print("Negative area:",
      (crop_clean["area_ha"] < 0).sum())

print("Negative production:",
      (crop_clean["production_tonnes"] < 0).sum())

print("Negative yield:",
      (crop_clean["yield_tonnes_ha"] < 0).sum())
```

```python
print(
    "Zero production:",
    (crop_clean["production_tonnes"] == 0).sum()
)

print(
    "Zero yield:",
    (crop_clean["yield_tonnes_ha"] == 0).sum()
)
```

```python
#Saving processed historical dataset
OUTPUT_PATH = (
    "../data/processed/crop_yield/"
    "des_apy_2013_14_2022_23.csv"
)

crop_clean.to_csv(
    OUTPUT_PATH,
    index=False
)

print("Saved:", OUTPUT_PATH)
```
